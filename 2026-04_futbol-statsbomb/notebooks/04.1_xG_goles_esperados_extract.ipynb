```python
df_competiciones_masculino["season_year_texto"] = df_competiciones_masculino["season_name"].astype(str).str[:4]

df_competiciones_masculino["season_year"] = pd.to_numeric(df_competiciones_masculino["season_year_texto"],errors="coerce")
```

```python
df_competiciones_modelo = df_competiciones_masculino[df_competiciones_masculino["season_year"] >= 2015].copy()

df_competiciones_modelo = df_competiciones_modelo.sort_values(["season_year", "competition_name"],ascending=[False, True])

df_competiciones_modelo
```

### 2) **PARTIDOS**

```python
df_competiciones_modelo = df_competiciones_modelo.reset_index(drop=True)

df_competiciones_modelo.head()
```

```python
lista_partidos = []

for i in range(len(df_competiciones_modelo)):
    
    competition_id = df_competiciones_modelo.loc[i, "competition_id"]
    season_id = df_competiciones_modelo.loc[i, "season_id"]
    competition_name = df_competiciones_modelo.loc[i, "competition_name"]
    season_name = df_competiciones_modelo.loc[i, "season_name"]
    season_year = df_competiciones_modelo.loc[i, "season_year"]
    country_name = df_competiciones_modelo.loc[i, "country_name"]
    
    print("Descargando:", competition_name, "-", season_name)
    
    df_partidos = sb.matches(
        competition_id=competition_id,
        season_id=season_id)
    
    df_partidos["competition_id"] = competition_id
    df_partidos["season_id"] = season_id
    df_partidos["competition_name"] = competition_name
    df_partidos["season_name"] = season_name
    df_partidos["season_year"] = season_year
    df_partidos["country_name"] = country_name
    
    lista_partidos.append(df_partidos)
```

```python
df_partidos_modelo = pd.concat(lista_partidos, ignore_index=True)

df_partidos_modelo.head()
```
